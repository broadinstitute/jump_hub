## Recent papers

### 2026-10-15 — JUMP-lite accepted to WACV 2027

**Conference paper · Accepted October 15, 2026 · WACV 2027 · Public
preprint available**

*JUMP-lite: Compact, reproducible benchmarking of cell representations*
has been accepted to WACV 2027. JUMP-lite provides a compact subset of
JUMP Cell Painting for developing and comparing cell-image
representations. The paper also introduces Nahual, an open-source
framework for reproducible model deployment.

Read the [public preprint](https://arxiv.org/abs/2608.07632), explore
the [JUMP-lite data and code](https://github.com/afermg/JUMP_lite), or
try [Nahual](https://github.com/afermg/nahual).

### 2025-06-09 — cp_measure accepted to CODEML at ICML 2025

**Workshop paper · Accepted June 9, 2025 · CODEML at ICML 2025
(non-archival) · Public preprint available**

*cp_measure: API-first feature extraction for image-based profiling
workflows* was accepted to the CODEML workshop at ICML 2025. The Python
library brings CellProfiler’s core measurement capabilities into a
modular, API-first tool, making morphological feature extraction easier
to integrate into automated image-analysis workflows.

The [public preprint](https://arxiv.org/abs/2507.01163) was posted on
July 1, 2025. Explore the [code and
documentation](https://github.com/afermg/cp_measure).
