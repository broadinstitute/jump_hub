## 2026-08-07 — JUMP-lite: compact, reproducible benchmarking of cell representations

**Preprint**

JUMP-lite provides a compact subset of JUMP Cell Painting for developing
and comparing cell-image representations. The paper also introduces
Nahual, an open-source framework for reproducible model deployment.

[arXiv](https://arxiv.org/abs/2608.07632) · [GitHub
(JUMP-lite)](https://github.com/afermg/JUMP_lite) · [GitHub
(Nahual)](https://github.com/afermg/nahual)

## 2025-09-15 — OASIS: multi-omics for chemical safety assessment

**Published · Toxicological Sciences**

The OASIS Consortium combines Cell Painting, transcriptomics, and
proteomics across cell models to improve chemical safety assessment. The
initiative aims to make safety testing more predictive of human biology
while reducing animal use.

[Paper](https://doi.org/10.1093/toxsci/kfaf128)

## 2025-08-07 — Morphological map of under- and overexpression of genes in human cells

**Published · Nature Methods**

The JUMP Cell Painting genetic dataset captures the effects of
perturbing 15,243 human genes through CRISPR knockouts and gene
overexpression. Its single-cell images and morphological profiles
support exploration of gene functions and relationships.

[Paper](https://doi.org/10.1038/s41592-025-02753-9)

## 2025-07-01 — cp_measure: API-first feature extraction for image-based profiling workflows

**Preprint**

cp_measure brings CellProfiler’s core measurement capabilities into a
modular, API-first Python library for morphological feature extraction.
It makes image-based profiling easier to integrate into automated
analysis workflows.

[arXiv](https://arxiv.org/abs/2507.01163) ·
[GitHub](https://github.com/afermg/cp_measure)

## 2025-03-03 — Reproducible image-based profiling with Pycytominer

**Published · Nature Methods**

Pycytominer is an open-source Python package for processing single-cell
measurements into profiles for downstream analysis. It provides
reproducible bioinformatics steps for image-based profiling workflows.

[Paper](https://doi.org/10.1038/s41592-025-02611-8)

## 2024-12-05 — Cell Painting: a decade of discovery and innovation in cellular imaging

**Published review · Nature Methods**

A review of a decade of Cell Painting advances, from assay protocols and
image analysis to biological applications. It covers feature extraction,
quality control, batch correction, and integration with other omics
data.

[Paper](https://doi.org/10.1038/s41592-024-02528-8)
